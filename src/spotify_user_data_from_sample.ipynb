## Your Spotify-style data (from sample files)

Uses **`data/sample_*.json`** — no API key needed. Same structure as the real Spotify API so you can test code now and switch to the API after Feb 11.

```python
import json
import os

# Path to data folder (run from repo root or from src/)
data_dir = os.path.join(os.getcwd(), 'data')
if not os.path.exists(data_dir):
    data_dir = os.path.join(os.getcwd(), '..', 'data')

with open(os.path.join(data_dir, 'sample_saved_albums.json'), encoding='utf-8') as f:
    saved_albums = json.load(f)
with open(os.path.join(data_dir, 'sample_liked_songs.json'), encoding='utf-8') as f:
    liked_songs = json.load(f)
with open(os.path.join(data_dir, 'sample_playlists.json'), encoding='utf-8') as f:
    playlists = json.load(f)

print('Loaded sample data from', data_dir)
```

```python
print("Your Saved Albums:")
for album in saved_albums['items']:
    a = album['album']
    artists = ', '.join(artist['name'] for artist in a['artists'])
    print(f" - {a['name']} by {artists}")

print("\nYour Liked Songs:")
for song in liked_songs['items']:
    t = song['track']
    artists = ', '.join(artist['name'] for artist in t['artists'])
    print(f" - {t['name']} by {artists}")

print("\nYour Playlists:")
for playlist in playlists['items']:
    print(f" - {playlist['name']} (ID: {playlist['id']})")
```
